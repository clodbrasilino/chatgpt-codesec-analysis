```python
import random
import numpy as np
from scipy import stats

def LLM_correction(questions, initial_errors):
    rounds = 0
    errors_left = initial_errors
    while errors_left > 0:
        rounds += 1
        for i, question in enumerate(questions):
            if question != "correta":
                # Simula a correção pelo LLM (atribui "correta" com uma probabilidade crescente a cada rodada)
                correction_probability = min(rounds * 0.01, 1.0)
                if random.random() < correction_probability:
                    questions[i] = "correta"
                    errors_left -= 1
    return rounds

def simulate(num_simulations, num_questions, initial_errors):
    total_rounds = []
    for _ in range(num_simulations):
        questions = ["incorreta"] * num_questions
        rounds = LLM_correction(questions, initial_errors)
        total_rounds.append(rounds)
    return total_rounds

num_questions = 974
num_simulations = 1000
initial_errors = 162.25  # Defina o número inicial de erros na primeira rodada

# Realizar as simulações
rounds_per_simulation = simulate(num_simulations, num_questions, initial_errors)

# Calcular a média de rodadas
average_rounds = np.mean(rounds_per_simulation)

print(f"Número médio de rodadas necessárias para corrigir todas as questões: {average_rounds}")
```

```python
# intervalo de confiança para a média considerando os percentis da simulação
percentile_5 = np.percentile(rounds_per_simulation, 5);print(percentile_5)
percentile_95 = np.percentile(rounds_per_simulation, 95);print(percentile_95)

print(f"Intervalo de confiança de 95% para o número médio de rodadas: {(percentile_5, percentile_95)}")
```

```python
# intervalo de confiança para a média considerando variância desconhecida
std_dev_rounds = np.std(rounds_per_simulation)
sample_size = len(rounds_per_simulation)

# erro padrão da média
standard_error = std_dev_rounds / np.sqrt(sample_size)

# valor crítico da distribuição t para um intervalo de confiança de 95%
t_critical = abs(stats.t.ppf((1 - 0.05 / 2), df=sample_size - 1))

confidence_interval = (average_rounds - t_critical * standard_error, average_rounds + t_critical * standard_error)

print(f"Intervalo de confiança de 95% para o número médio de rodadas: {confidence_interval}")
```
